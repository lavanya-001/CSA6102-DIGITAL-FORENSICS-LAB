```python
def recover_deleted_messages(sms_table):
    """
    sms_table: List of dictionaries with:
    - is_deleted (bool)
    - overwritten (bool)

    Returns a list of deleted but recoverable messages.
    """

    recoverable = [
        row
        for row in sms_table
        if row["is_deleted"] and not row["overwritten"]
    ]

    return recoverable


def summarize_table(sms_table):
    """
    Returns a summary of SMS records.
    """

    active = [
        r
        for r in sms_table
        if not r["is_deleted"]
    ]

    recoverable = recover_deleted_messages(sms_table)

    lost = [
        r
        for r in sms_table
        if r["is_deleted"] and r["overwritten"]
    ]

    return {
        "active": len(active),
        "recoverable": len(recoverable),
        "permanently_lost": len(lost),
    }
```

```python
sms_table = [
    {
        "id": 1,
        "message": "Hello",
        "is_deleted": False,
        "overwritten": False
    },
    {
        "id": 2,
        "message": "Secret Message",
        "is_deleted": True,
        "overwritten": False
    },
    {
        "id": 3,
        "message": "Old SMS",
        "is_deleted": True,
        "overwritten": True
    },
    {
        "id": 4,
        "message": "Meeting at 5 PM",
        "is_deleted": False,
        "overwritten": False
    }
]

print(recover_deleted_messages(sms_table))
print(summarize_table(sms_table))
```
