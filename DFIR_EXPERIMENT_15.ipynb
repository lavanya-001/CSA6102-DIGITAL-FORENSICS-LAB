```python
import hashlib
import pandas as pd

# -------------------------------
# Sample Files
# -------------------------------
files = {
    "invoice.pdf": b"Hello World",
    "update.exe": b"This is a malware sample",
    "photo.jpg": b"My holiday image",
    "trojan.exe": b"This is a malware sample"
}

# -------------------------------
# Generate MD5, SHA1, SHA256
# -------------------------------
data = []

for filename, content in files.items():
    md5 = hashlib.md5(content).hexdigest()
    sha1 = hashlib.sha1(content).hexdigest()
    sha256 = hashlib.sha256(content).hexdigest()

    data.append([filename, md5, sha1, sha256])

df = pd.DataFrame(data, columns=["File Name", "MD5", "SHA1", "SHA256"])

# Save Malware Hash Database
df.to_csv("malware_hash_database.csv", index=False)

print("=== Malware Hash Database ===")
print(df)

# -------------------------------
# Simple Hash-Based Detection
# -------------------------------
known_malware = set(df[df["File Name"].isin(["update.exe", "trojan.exe"])]["SHA256"])

print("\n=== Detection Results ===")

detected = 0

for i in range(len(df)):
    if df.loc[i, "SHA256"] in known_malware:
        status = "Malware Detected"
        detected += 1
    else:
        status = "Safe File"

    print(df.loc[i, "File Name"], "->", status)

# -------------------------------
# Detection Statistics
# -------------------------------
print("\n=== Detection Rate Statistics ===")
print("Total Files:", len(df))
print("Detected Malware:", detected)
print("Safe Files:", len(df) - detected)
print("Detection Rate: {:.2f}%".format((detected/len(df))*100))

# -------------------------------
# Hash Family Clustering
# -------------------------------
print("\n=== Hash Family Clustering ===")

clusters = {}

for sha in df["SHA256"]:
    prefix = sha[:8]
    clusters.setdefault(prefix, 0)
    clusters[prefix] += 1

for key, value in clusters.items():
    print(key, ":", value, "file(s)")

# -------------------------------
# Hash Collision Demonstration
# -------------------------------
print("\n=== Hash Collision Analysis ===")
print("MD5 and SHA1 have known collision vulnerabilities.")
print("SHA256 is currently recommended for malware identification.")

print("\nCSV file 'malware_hash_database.csv' created successfully.")
```
