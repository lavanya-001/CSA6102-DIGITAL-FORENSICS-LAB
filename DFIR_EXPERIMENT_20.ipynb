```python
import os

# Create source folder
os.makedirs("acquisition_demo/source", exist_ok=True)
source = "acquisition_demo/source"

# Create sample files
with open(os.path.join(source, "report.docx"), "w") as f:
    f.write("Confidential quarterly report.")

with open(os.path.join(source, "photo.jpg"), "w") as f:
    f.write("FAKEJPEGDATA")

# Simulated raw disk image
raw_disk = (
    b"REPORTDOCX_CONTENT"
    + b"\x00" * 20
    + b"DELETED_INVOICE_DATA"
    + b"\x00" * 10
)

# Physical acquisition
def physical_acquisition(raw_bytes):
    return bytes(raw_bytes)   # Every bit: used + unused + deleted

# Logical acquisition
def logical_acquisition(folder):
    return {
        f: open(os.path.join(folder, f), "rb").read()
        for f in os.listdir(folder)
    }

# Sparse acquisition
def sparse_acquisition(folder, targets):
    return {
        f: open(os.path.join(folder, f), "rb").read()
        for f in targets
        if f in os.listdir(folder)
    }

# Test the functions
physical = physical_acquisition(raw_disk)
logical = logical_acquisition(source)
sparse = sparse_acquisition(source, ["report.docx"])

print("Physical Acquisition Size:", len(physical), "bytes")
print("Logical Acquisition Files:", list(logical.keys()))
print("Sparse Acquisition Files:", list(sparse.keys()))
```
