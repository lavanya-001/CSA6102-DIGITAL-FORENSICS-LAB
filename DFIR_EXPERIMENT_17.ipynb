```python
import pandas as pd

# ----------------------------------------
# Sample Login Attempts (Brute Force Simulation)
# ----------------------------------------

login_attempts = [
    {"Username": "admin", "Password": "123456"},
    {"Username": "admin", "Password": "password"},
    {"Username": "admin", "Password": "admin123"},
    {"Username": "admin", "Password": "Secure@123"}
]

correct_password = "Secure@123"

print("========== BRUTE FORCE ATTACK SIMULATION ==========")

logs = []

for attempt in login_attempts:
    if attempt["Password"] == correct_password:
        status = "SUCCESS"
    else:
        status = "FAILED"

    logs.append([attempt["Username"], attempt["Password"], status])
    print(attempt["Username"], "-", attempt["Password"], "->", status)

# ----------------------------------------
# Weak Password Analysis
# ----------------------------------------

print("\n========== WEAK PASSWORD ANALYSIS ==========")

weak_passwords = ["123456", "password", "admin123", "qwerty"]

for pwd in weak_passwords:
    print(pwd, "-> Weak Password")

# ----------------------------------------
# SQL Injection Detection
# ----------------------------------------

print("\n========== SQL INJECTION DETECTION ==========")

inputs = [
    "admin",
    "' OR '1'='1",
    "John",
    "'; DROP TABLE users;--"
]

sql_keywords = ["OR", "DROP", "--", "'"]

for text in inputs:
    detected = any(keyword.lower() in text.lower() for keyword in sql_keywords)

    if detected:
        print(text, "-> Possible SQL Injection")
    else:
        print(text, "-> Normal Input")

# ----------------------------------------
# Directory Traversal Detection
# ----------------------------------------

print("\n========== DIRECTORY TRAVERSAL DETECTION ==========")

paths = [
    "/images/logo.png",
    "../../etc/passwd",
    "/home/user/file.txt",
    "../secret/config.txt"
]

for path in paths:
    if "../" in path:
        print(path, "-> Directory Traversal Detected")
    else:
        print(path, "-> Safe Path")

# ----------------------------------------
# Attack Pattern Logs
# ----------------------------------------

df = pd.DataFrame(logs, columns=["Username", "Password", "Result"])

df.to_csv("attack_pattern_logs.csv", index=False)

print("\nAttack logs saved as attack_pattern_logs.csv")

# ----------------------------------------
# Vulnerability Classification
# ----------------------------------------

print("\n========== VULNERABILITY CLASSIFICATION ==========")

print("1. Weak Passwords - HIGH")
print("2. SQL Injection Attempts - HIGH")
print("3. Directory Traversal Attempts - MEDIUM")
print("4. Brute Force Attempts - HIGH")

# ----------------------------------------
# WAF Rule Suggestions
# ----------------------------------------

print("\n========== WAF RULE SUGGESTIONS ==========")

rules = [
    "Block repeated failed login attempts.",
    "Filter SQL keywords such as OR, DROP, UNION.",
    "Reject '../' in URL requests.",
    "Enable CAPTCHA after multiple login failures.",
    "Use strong password policy."
]

for i, rule in enumerate(rules, 1):
    print(f"{i}. {rule}")
```
