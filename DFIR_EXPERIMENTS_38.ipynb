```python
import hashlib
import math
from collections import Counter


def sha256_hash(data: bytes) -> str:
    """
    Compute the SHA-256 hash of the given bytes.
    """
    return hashlib.sha256(data).hexdigest()


def byte_entropy(data: bytes) -> float:
    """
    Calculate the Shannon entropy of byte data.
    """

    if not data:
        return 0.0

    counts = Counter(data)
    length = len(data)

    return -sum(
        (count / length) * math.log2(count / length)
        for count in counts.values()
    )


def static_analyze(data: bytes, entropy_threshold=7.5):
    """
    Perform simple static analysis on binary data.

    Returns:
    - SHA-256 hash
    - File size
    - Entropy
    - Whether the file is likely packed/encrypted
    """

    entropy = byte_entropy(data)

    return {
        "sha256": sha256_hash(data),
        "size_bytes": len(data),
        "entropy": round(entropy, 2),
        "likely_packed": entropy >= entropy_threshold,
    }
```

```python
sample_data = b"This is a sample malware file used for testing."

result = static_analyze(sample_data)

print(result)
```
